```python
# n = int(input())
n = 6
for i in range(2*n):
    if i % 2 ==0:
        # i=0; 1
        # i=2; 2
        # i=4; 3
        # i=6; 4
        # i=8; 5
        idx = int((i+2)/2)
        for j in range(idx):
            print('*', end=' ')
    else:
        # i=1; 5
        # i=3; 4
        # i=5; 3
        # i=7; 2
        # i=9; 1
        idx = int((2*n-i)/2)
        for k in range(idx, -1, -1):
            print('*', end=' ')
    print()
```

```python
n = 3

for i in range(2*n+1): # i=0,1,2,...,6
    for j in range(2*n+1):
        if i % 2 == 0 or j % 2 == 0:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
```

```python
arr = [1, 2, 4, 5, 7, 8, 9, 10, 0, 3]
cnt = 0
sum = 0
for i in arr:
    if i == 0:
        break

    if i % 2 == 0:
        cnt += 1
        sum += i

print(cnt, sum)
```

```python
arr = [[10, 20, 30, 40],
       [50, 60, 70, 80]]
arr
```

```python
print(arr)

ga = len(arr) # 행
se = len(arr[0]) # 열

# 가로 평균
for i in range(ga):
    print(f'{sum(arr[i])/se:.1f}', end=' ')

# 세로 평균
print()
for j in range(se):
    temp = 0
    for i in range(ga):
        temp += arr[i][j]
    print(f'{temp/ga:.1f}', end=' ')
# 전체 평균
result = 0
for i in arr:
    result += sum(i)

print()
print(result/(ga*se))

# print(sum(arr[0]))
# sum(arr[0])
```
