```python
# 2. 각 연도별 최대 검거율을 가진 범죄유형을 찾아서 해당 연도 및 유형의 검거건수들의 총합을 구하시오. (검거율 = 검거건수 / 발생건수)
# data
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_1_2.csv')
df.head()
```

```python
df1 = df[df['구분'] == '발생건수'].set_index('연도').drop(columns='구분')
df2 = df[df['구분'] == '검거건수'].set_index('연도').drop(columns='구분')
dfratio = df2 / df1
dfratio.head()
```

```python
max_per_year = dfratio.max(axis=1)
max_per_year
```

```python
def is_max_col(col):
    return col == max_per_year

mask = dfratio.apply(is_max_col, axis = 0)
mask
```

```python
df2_1 = df2[mask].fillna(0)
df2_1.sum().sum()
```

```python
# data
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_2_train.csv')
df
```

```python
train_X = df.drop(columns='productivity')
train_X
train_y = df['productivity']
train_y
```

```python
import statsmodels.api as sm
train_X=sm.add_constant(train_X)
```

```python
model = sm.OLS(train_y, train_X).fit()
model.summary()
```
