```python
# 2. 제시된 문제를 순서대로 풀고, 해답을 제시하시오.
# data
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_1_2.csv')
df
```

```python
# 2.1 date를 연도(year), 월(month)로 분리하여, 연도-월별 price의 합계를 구하시오. 그 중 두번째로 큰 매출액(합계)을 구하시오.
df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df
```

```python
# 2.2 이전 문제에서 네 번째로 큰 price 합계에 해당하는 연도-월을 찾으시오. 해당 연도-월에서 카테고리별 price 합계를 구하시오. 그 중 가장 높은 price 합계(정수)를 제출하시오.
monthly_sales = price.sort_values(ascending=False)
year, month = monthly_sales.reset_index().iloc[3][['year', 'month']]
df_sub = df.loc[(df['year'] == year) & (df['month'] == month), :]
df_sub.groupby('category')['price'].sum().max()
```

# 제2유형

```python
# 제공된 학습용 데이터(10_2_train.csv)는 여러 상권 내 건물의 특성(상권 유형, 건물 면적, 건물 연식, 세대 수 등)과 연간 총 가스 소비량 정보를 담고 있다. 
# 학습용 데이터를 활용하여 건물의 연간 총 가스 소비량(gas_totl)을 예측하는 모델을 개발하고, 이 중 가장 우수한 모델을 평가용 데이터(10_2_test.csv)에 적용하여 예측 결과를 제출하시오.
# 평가 지표: RMSE
```

```python
# data
train = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_2_train.csv')
test = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_2_test.csv')
train
```

```python
print(train.info())
print(test.info())
```

```python
# 결측값 제거
train = train.loc[train['gas_totl'] != 0, :]
train_X = train.drop(columns='gas_totl', axis=1)
train_y = train['gas_totl']

test_X = test.drop(columns='gas_totl', axis=1)
test_y = test['gas_totl']
```

```python
print(train_X.shape, train_y.shape, test_X.shape, test_y.shape)
```

```python
# data split
from sklearn.model_selection import train_test_split
train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3, random_state=1)
print(train_X.shape, train_y.shape, valid_X.shape, valid_y.shape)
```

```python
# 데이터 전처리 - 원핫인코딩
from sklearn.preprocessing import OneHotEncoder
cat_columns = train_X.select_dtypes('object').columns
num_columns = train_X.select_dtypes('number').columns
print(cat_columns)
print(num_columns)
```

```python
encoder = OneHotEncoder(sparse_output=False).set_output(transform='pandas')

train_X_cat_oh = encoder.fit_transform(train_X[cat_columns])
valid_X_cat_oh = encoder.transform(valid_X[cat_columns])
test_X_cat_oh = encoder.transform(test_X[cat_columns])

train_X_cat_oh
```

```python
train_X_preprocessed = pd.concat(
    [train_X_cat_oh, train_X[num_columns]], axis=1
)
valid_X_preprocessed = pd.concat(
    [valid_X_cat_oh, valid_X[num_columns]], axis=1
)
test_X_preprocessed = pd.concat(
    [test_X_cat_oh, test_X[num_columns]], axis=1
)
train_X_preprocessed
```

```python
# 랜덤 포레스트
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(random_state=1)
rf.fit(train_X_preprocessed, train_y)

```

```python
from sklearn.metrics import root_mean_squared_error
pred = rf.predict(valid_X_preprocessed)

rmse = root_mean_squared_error(valid_y, pred)
rmse
```

```python
# test 데이터로 최종 예측 수행
test_pred = rf.predict(test_X_preprocessed)
test_pred = pd.DataFrame(test_pred, columns=['pred'])
test_pred.head()

# 저장
test_pred.to_csv('result.csv', index=False)
```

# 제3유형

```python
# 1. 한 기업이 인사관리 데이터를 가지고 이직 여부 예측 모델을 개발하려고 한다. 다음 문제를 풀이하시오.
# data
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_3_1.csv')
df
```

```python
# 1.1 이직 여부를 예측하는 로지스틱 회귀모형을 적합하고, 유의한 변수(유의확률 0.05 미만)의 회귀계수를 소수점 셋째 자리까지 반올림하여 제출하시오. (단, 절편 제외)
df.info()
df.describe()
```

```python
import statsmodels.formula.api as smf
model = smf.logit('attrition~age+income+C(overtime)', data=df).fit()
pvalues = model.pvalues[1:]
params = model.params[1:]
print(pvalues)
print()
print(params)
```

```python
# 유의수준 0.05에서 income 변수의 pvalue가 0.05 이하이므로 유의.
sig_var = params[pvalues < 0.05]
res = np.round(sig_var, 3)
res
```

```python
# 1.2 age가 1 증가할 때 이직(또는 잔류) 오즈비(odds ratio)를 소수점 셋째 자리까지 반올림하여 제출하시오.
age_coef = model.params['age']
odds_ratio_age = np.exp(age_coef)
odds_ratio_age_rounded = round(odds_ratio_age, 3)
print(odds_ratio_age_rounded)
```

```python
# 1.3 age=20, income=3000, overtime=2 값을 가진 데이터의 이직확률을 모델로 예측하여 소수점 셋째 자리까지 반올림하여 제출하시오.
test_df = pd.DataFrame({'age':[20],
              'income':[3000],
              'overtime':[2]})
test_df
```

```python
# 예측 확률 산출
pred_prob = model.predict(test_df)
np.round(pred_prob, 3)

# # 직접 계산
# a = model.params['Intercept'] + model.params['age'] * 20 + model.params['income'] * 3000 +  model.params['C(overtime)[T.2]']
# np.exp(a)/(1+np.exp(a))
```

```python
# 2. 어느 지역의 주택들의 정보를 수집하여 주택 가격을 예측하는 모델을 개발하려고 한다.
# data
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_3_2.csv')
df
```

```python
df.info()
df.describe()
```

```python
# 2.1 주택 가격을 예측하는 다중선형회귀모형을 적합하고, 유의한 변수(유의확률 0.05 미만)의 회귀계수 합(절편 제외)을 소수점 셋째 자리까지 반올림하여 제출하시오.
import statsmodels.formula.api as smf

# 다중회귀적합
model = smf.ols('price~area+height+wall', data=df).fit()
params = model.params[1:]
pvalues = model.pvalues[1:]

sig_var = params[pvalues < 0.05]
ceof_sum = sig_var.sum()
np.round(ceof_sum, 3)
```

```python
sig_var
```

```python
# 2.2 유의한 변수만으로 다중선형회귀모형을 다시 적합하고, 결정계수를 소수점 셋째 자리까지 반올림하여 제출하시오.
model = smf.ols('price ~ area +height', data=df).fit()
np.round(model.rsquared, 3)

# # 또는
# sig_var_names = sig_var.index.tolist()
# formula = 'price ~ ' + ' + '.join(sig_var_names)
# 이하 동일
```

```python
# 2.3 area=100, height=10, wall=1 값을 가진 데이터의 예측 주택 가격을 모델로 예측하여 소수점 셋째 자리까지 반올림하여 제출하시오. (단, 이전 문제에서 뽑은 통계적으로 유의미한 변수만 선택할 것)
```

```python
test_df = pd.DataFrame({
    'area' : [100],
    'height' : [10],
    'wall' : [1]
})
sig_var_names = sig_var.index.tolist()
test_X = test_df[sig_var_names]

# 예측
pred = model.predict(test_X)
np.round(pred, 3)
```
