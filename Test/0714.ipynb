```python
from scipy import stats

ctr = [92.45, 91.92, 92.64, 93.87, 92.69]
cats = [93.38, 93.67, 92.76, 94.7, 92.93]

t_statistic, p_value = stats.ttest_rel(ctr, cats)

print('대응표본 t-검정 결과:')
print(f't-통계량 : {t_statistic}')
print(f'p-값 : {p_value}')
```
