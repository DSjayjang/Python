```python
# 1. 상삼각행렬 마스크 생성 (대각선과 하삼각 영역을 제외하기 위함)
# k=1 설정을 통해 대각선(자기자신) 바로 위부터 True로 만듭니다.
mask = np.triu(np.ones(df_corr.shape), k=1).astype(bool)
mask
```

```python
# 2. 마스크를 적용하여 대각선 및 하삼각 영역을 NaN으로 채우기
upper_tri = df_corr.where(mask)

# 3. 2차원 표 형태를 1차원 시리즈로 압축(stack)하면서 NaN은 자동으로 제거
stacked_corr = upper_tri.stack()

# 4. 값이 0.9 이상인 컬럼 쌍만 필터링
high_corr_pairs = stacked_corr[stacked_corr >= 0.9]
print(high_corr_pairs)
```

```python
# dataset
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/dataset/main/Parkinsons.csv')
df
```

```python
# 'name' 제거
df = df.drop(columns=['name'], axis=1)
df.shape
```

```python
from sklearn.preprocessing import minmax_scale

df_scaled = minmax_scale(df)
df_scaled = pd.DataFrame(df_scaled)
df_scaled
```

```python
df_scaled.columns = df.columns
df_scaled
```

```python
import statsmodels.api as sm
df_scaled = sm.add_constant(df_scaled, has_constant='add')
df_scaled
```

```python
feature_columns = list(df_scaled.columns.difference(['status']))

X = df_scaled[feature_columns]
y = df_scaled['status'].astype('category')
```

```python
# data split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y,  test_size=0.1, random_state=1)
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
import statsmodels.api as sm
model = sm.Logit(y_train, X_train)

results = model.fit(method='bfgs', maxiter=1000)
results.summary()
```

```python
np.sort(results.pvalues)
```

```python
results.pvalues.sort_values()
```

```python
def cut_off(y, threshold):
    Y = y.copy()
    Y[Y>threshold] = 1
    Y[Y<=threshold] = 0
    return(Y.astype(int))
```

```python
y_pred_prob = results.predict(X_test)
y_pred = cut_off(y_pred_prob, 0.8)
y_pred
```

```python
# f1 score 계산
from sklearn.metrics import f1_score
f1_score(y_test, y_pred)
```

```python
# scaling
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_train_scaling = scaler.fit_transform(X_train)
X_test_scaling = scaler.fit_transform(X_test)

```

```python
# logistic regression
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(solver='bfgs')
model.fit(X_train_scaling, y_train)
model
LogisticRegression()
```

```python
y_pred = model.predict(X_test_scaling)
y_pred
```

```python

f1_score(y_test, y_pred)
```

```python
# dataset
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/dataset/main/HR-Employee-Attrition.csv')
df
```

```python
df[['MaritalStatus', 'OverTime']]
```

```python
table = pd.crosstab(df['OverTime'], df['MaritalStatus'])
table
```

```python
# X1: 초과근로자에 대한 결혼 여부
# X2: 초과근로하지 않은 자의 결혼 여부
X1 = table.loc['Yes', :]
X2 = table.loc['No', :]

print(X1)
print(X2)
```

```python
X1_married = X1['Married']
X2_married = X2['Married']

print(abs(X1_married - X2_married))
```

```python
from scipy.stats import chi2_contingency
result = chi2_contingency([X1, X2])
print(round(result.statistic, 2))
```

```python
p_value = result.pvalue
print(round(p_value, 4))

if p_value < 0.05:
    print('기각')
else:
    print('채택')
```
